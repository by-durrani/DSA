Numerical Hash

``````python
def hash():
    size_of_arr = int(input("size of array"))
    
    # create a fix size of array
    arr_of_numbers = [None] * size_of_arr
    
    # insert elements in fixed array 
    for i in range(size_of_arr):
        arr_of_numbers[i] = int(input("insert element of array"))
    
    # precompute
    hash = [0] * (size_of_arr + 1)
    for i in range(size_of_arr):
        hash[arr_of_numbers[i]] += 1
        
    # limit of checking numbers list
    q = int(input("limit of list you want to ask"))
    while q > 0 :
        # fetching
        x = int(input("which element you want to check?"))
        print(hash[x])
        # updating local state
        q -= 1
    
hash()
``````

Alphabetical Hash with 26 max length (character hashing)

``````python
def hash():
    string = input("insert element of array")
    
    # hash array of alphabets
    hash = [0] * 26
    for i in range(len(string)):
        hash[ord(string[i]) - ord('a')] += 1
        
    # limit of checking numbers list
    q = int(input("limit of list you want to ask"))
    while q > 0 :
        x = int(ord(input("which alphabet you want to check?")))
        # fetching
        chr_idx = hash[x - ord("a")]
        print(chr_idx)
        # updating local state
        q -= 1
    
hash()
``````
